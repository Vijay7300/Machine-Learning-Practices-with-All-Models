```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('C:/COLLEGE DATA/College profile/Machine Learning/ML_Practice_with_python_code/Datasets/Placement.csv')
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel("Package")
```

```python
X=df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
X_train ,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)

lr=LinearRegression() # this is object for LR

lr.fit(X_train,y_train)

lr.predict(X_test.iloc[1].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('package in lpa')
```

```python
# by mathematically
m=lr.coef_
c=lr.intercept_
```

```python
# formula  y=mx+c
# let suppose x= 7 cgpa then y=? in lpa
x=7
y=m*x+c
y
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
y_pred=lr.predict(X_test)
```

```python
print("MSE",mean_squared_error(y_test,y_pred)) # not in LPA^2
print("MAE",mean_absolute_error(y_test,y_pred))  # in LPA
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))  # in LPA
```

```python
# Coefficient of Determination
print("R2_score",r2_score(y_test,y_pred))
```

```python
n=len(X_test)
n # no.of rows/samples
```

```python
# AD_R2_score is better for more no.of features & irrelavent features

k=1 # Number of independent (input) features
r2 = r2_score(y_test,y_pred)
Ad_R2= 1-(((1-r2)*(n-1))/(n-1-k))

print("Adj_R2_score",Ad_R2)
```
