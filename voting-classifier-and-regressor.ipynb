The Iris dataset has three relatively distinct clusters, and petal length and petal width provide the strongest visual separation between the species. Therefore, the dataset should be relatively easy for classification models such as Decision Tree, KNN, Logistic Regression, and Voting Ensemble.

```python
new_df = df[df['Species'] != 0][['SepalLengthCm','SepalWidthCm','Species']]
new_df
```

## Now perform ML Models on this dataset

```python
X=new_df.iloc[:,0:2]
y=new_df.iloc[:,-1]
```

```python
clf1=LogisticRegression()
clf2=RandomForestClassifier()
clf3=KNeighborsClassifier()
```

```python
estimators=[('lr: ',clf1),('rf: ',clf2),('knn: ',clf3)]
```

```python
for estimator in estimators:
    x=cross_val_score(estimator[1],X,y,cv=10,scoring='accuracy')
    print(estimator[0],np.mean(x))
```

## Hard Voting

```python
vc=VotingClassifier(estimators=estimators,voting='hard')
x=cross_val_score(vc,X,y,cv=10,scoring='accuracy')
print(np.mean(x))
```

## Soft Voting

```python
vc=VotingClassifier(estimators=estimators,voting='soft')
x=cross_val_score(vc,X,y,cv=10,scoring='accuracy')
print(np.mean(x))
```

## Weighted Voting

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingClassifier(estimators=estimators,voting='soft',weights=[i,j,k])
            x = cross_val_score(vc,X,y,cv=10,scoring='accuracy')
            print("for i={},j={},k={}".format(i,j,k),np.round(np.mean(x),2))
```

## Classifiers of Same Algorithm

```python
from sklearn.datasets import make_classification
from sklearn.calibration import CalibratedClassifierCV
```

```python
X,y=make_classification(n_samples=1000,n_features=20,n_informative=15,random_state=2,n_redundant=5)

svm1 = CalibratedClassifierCV(
    SVC(kernel='poly', degree=1),
    ensemble=False)

svm2 = CalibratedClassifierCV(
    SVC(kernel='poly', degree=2),
    ensemble=False)

svm3 = CalibratedClassifierCV(
    SVC(kernel='poly', degree=3),
    ensemble=False)

svm4 = CalibratedClassifierCV(
    SVC(kernel='poly', degree=4),
    ensemble=False)

svm5 = CalibratedClassifierCV(
    SVC(kernel='poly', degree=5),
    ensemble=False)



estimators = [('svm1',svm1),('svm2',svm2),('svm3',svm3),('svm4',svm4),('svm5',svm5)]

print("Individual SVM Performance:")

for estimator in estimators:
    x = cross_val_score(estimator[1],X,y,cv=10,scoring='accuracy')
    print(estimator[0],np.round(np.mean(x),2))


# 5. Create Voting Ensemble
voting = VotingClassifier(
    estimators=estimators,
    voting='soft'
)


# 6. Evaluate Voting Ensemble
voting_scores = cross_val_score(
    voting,
    X,
    y,
    cv=10,
    scoring='accuracy'
)

print("\nVoting Ensemble Performance:")

print(
    "Voting Classifier:",
    np.round(np.mean(voting_scores), 2)
)
```

# **Regressor**

```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor,VotingRegressor
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.datasets import make_regression

```

```python
X,y=make_regression(n_samples=500,n_features=20,n_informative=15,random_state=2)
```

```python
lr=LinearRegression()
rf=RandomForestRegressor()
svr=SVR()
```

```python
estimators=[('lr',lr),('rf',rf),('svr',svr)]
```

```python
for estimator in estimators:
    x=cross_val_score(estimator[1],X,y,cv=10,scoring='r2')
    print(estimator[0],np.mean(x))
```

```python
vr=VotingRegressor(estimators=estimators)
x=cross_val_score(vr,X,y,cv=10,scoring='r2')
print('Voting_regressor: ',np.mean(x))
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vr=VotingRegressor(estimators,weights=(i,j,k))
            x=cross_val_score(vr,X,y,cv=10,scoring='r2')
            print('for i={},j={},k={}'.format(i,j,k),np.mean(x))
```
