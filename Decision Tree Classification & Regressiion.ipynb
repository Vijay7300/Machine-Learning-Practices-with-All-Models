```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import accuracy_score,r2_score
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
```

## DT For Classifiaction

```python
df=load_iris()
```

```python
X=df['data']
y=df['target']
```

```python
X.shape
```

```python
y.shape
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=43)
```

```python
dclf=DecisionTreeClassifier()
dclf.fit(X_train,y_train)
y_pred=dclf.predict(X_test)
```

```python
dclf
```

```python
print("accuracy",accuracy_score(y_test,y_pred))
```

## DT Visulazation

```python
from sklearn.tree import plot_tree
from matplotlib.pylab import rcParams
```

```python
rcParams['figure.figsize']=80,50
plot_tree(dclf)
```

## DT For Regression

```python
df1=pd.read_csv('C:\COLLEGE DATA\College profile\Machine Learning\ML_Practice_with_python_code\Datasets\placement.csv')
df1.head()
```

```python
X1=df1[['cgpa']]
y1=df1[['package']]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X1,y1,test_size=0.2,random_state=4)
```

```python
dtr=DecisionTreeRegressor(criterion='squared_error',max_depth=3)
dtr.fit(X_train,y_train)
y_pred1=dtr.predict(X_test)
```

```python
print("r2_score",r2_score(y_test,y_pred1))
```

### Hyperparameter Tuning
GridSearchCV is a hyperparameter tuning technique that exhaustively tests combinations of specified hyperparameters using cross-validation and returns the combination that gives the best validation performance.

```python
param_grid={'max_depth':[2,4,6,8,None],
            'criterion':['squared_error','absolute_error'],
            'max_features':[0.25,0.5,0.75,1.0],
            'min_samples_split':[0.25,0.5,0.75,1.0]}
```

```python
reg=GridSearchCV(DecisionTreeRegressor(),param_grid=param_grid)
reg.fit(X_train,y_train)
```

```python
reg.best_score_
```

```python
reg.best_params_
```

GridSearchCV matters because it helps select optimal hyperparameters and can improve the model's generalization performance while reducing the risk of overfitting.
