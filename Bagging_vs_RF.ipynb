```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
  
```

```python

df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

```python
bag=BaggingClassifier(max_features=2)
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

```python
rf=RandomForestClassifier(max_features=2)
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
     
```

## Conclusion: 
### Bagging vs Random Forest

* **Bagging:** Can use multiple base algorithms such as DT, KNN, SVM, etc.
* **Random Forest:** Uses only **Decision Trees**.
* **Bagging:** Mainly performs **row sampling** during the sampling columns are fixed initially during whole tree.
* **Random Forest:** Performs **row sampling + feature sampling at each node** change columns on every node during sampling.
* Therefore, Random Forest introduces **more randomness** and generally gives better performance.

### Important Parameters

* `max_features` → Number of columns/features
* `max_samples` → Number of rows/samples
* `n_estimators` → Number of trees
* `bootstrap` → Random sampling of rows
* `oob_score` → Out-of-Bag evaluation
* `n_jobs` → Parallel processing

**Feature importance = Column importance**
