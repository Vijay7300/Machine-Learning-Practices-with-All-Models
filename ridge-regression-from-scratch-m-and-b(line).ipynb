```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
X,y=make_regression(n_informative=1,n_samples=100,n_targets=1,random_state=13,n_features=1,noise=20)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression,Ridge
```

```python
lr=LinearRegression()
lr.fit(X,y)

print(lr.coef_)
print(lr.intercept_)
```

```python
R1=Ridge(alpha=1)
R1.fit(X,y)

print(R1.coef_)
print(R1.intercept_)
```

```python
R2=Ridge(alpha=10)
R2.fit(X,y)

print(R2.coef_)
print(R2.intercept_)
```

```python
R3=Ridge(alpha=100)
R3.fit(X,y)

print(R3.coef_)
print(R3.intercept_)
```

```python
plt.plot(X,y,'b.')
plt.plot(X,lr.predict(X),color='red',label='alpha=0')
plt.plot(X,R1.predict(X),color='green',label='alpha=1')
plt.plot(X,R2.predict(X),color='orange',label='alpha=10')
plt.plot(X,R3.predict(X),color='blue',label='alpha=100')
plt.legend()
```
